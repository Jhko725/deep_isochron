# INN schematics for the KSIAM talk

This notebook writes the two schematics of slides 9a and 9b to `results/ksiam/`, as PDF and PNG. The outputs are not committed. Labels and font sizes are fields of `StackStyle` and `WarpStyle`; change one with `copy.replace`.

## Setup

```python
from pathlib import Path

import jax
import matplotlib.pyplot as plt
from deep_isochron.plotting import (
    configure_rcparams,
    plot_coupling_warps,
    plot_inn_stack,
    schematic_affine_coupling,
    schematic_linear,
    schematic_spline_coupling,
    schematic_trajectories,
)


jax.config.update("jax_enable_x64", True)
configure_rcparams()
out = Path("results/ksiam")
out.mkdir(parents=True, exist_ok=True)


def save(fig, name):
    """Write `fig` as PDF and PNG under `out`."""
    for ext in ("pdf", "png"):
        fig.savefig(out / f"{name}.{ext}", bbox_inches="tight")
```

## 9a: the network as a chain of linear and spline coupling layers

Six blocks with an ellipsis, `H` above and `H⁻¹` below, and an inset under each block type, in `s9a_inn_stack.pdf`.

```python
fig = plot_inn_stack(schematic_linear(), schematic_spline_coupling())
save(fig, "s9a_inn_stack")
plt.show()
```

## 9b: the spline coupling layer beside an affine foil

Columns identity | affine | spline, rows the warp of each column and the grid with Bautin trajectories before and after the layer, in `s9b_coupling_warps.pdf`.

```python
fig = plot_coupling_warps(
    schematic_affine_coupling(),
    schematic_spline_coupling(),
    schematic_trajectories((-1.0, 1.0)),
)
save(fig, "s9b_coupling_warps")
plt.show()
```
